```python
import os
import pandas as pd

INPUT_FOLDER = r"3. Output"
OUTPUT_FOLDER = r"harmonized_sessions"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

DATASETS = {
    "acn.csv": {
        "source_dataset": "acn",
        "region": "California"
    },
    "palo_alto.csv": {
        "source_dataset": "palo_alto",
        "region": "California"
    },
    "boulder_2021.csv": {
        "source_dataset": "boulder",
        "region": "Colorado"
    },
    "dundee.csv": {
        "source_dataset": "dundee",
        "region": "UK"
    },
    "perth.csv": {
        "source_dataset": "perth",
        "region": "Australia"
    },
    "paris.csv": {
        "source_dataset": "paris",
        "region": "France"
    },
    "sap.csv": {
        "source_dataset": "sap",
        "region": "France"
    }
}

for file, meta in DATASETS.items():

    print(f"\nProcessing {file}")

    df = pd.read_csv(os.path.join(INPUT_FOLDER, file))

    df = df.rename(
        columns={
            "Start": "connect_time_utc",
            "End": "charge_end_time_utc",
            "Energy": "energy_kwh"
        }
    )

    df["connect_time_utc"] = pd.to_datetime(df["connect_time_utc"])
    df["charge_end_time_utc"] = pd.to_datetime(df["charge_end_time_utc"])

    df["source_dataset"] = meta["source_dataset"]

    if file == "acn.csv":

        df["entity_id"] = df["data"]
        df["site"] = df["data"]

    else:

        name = meta["source_dataset"]

        df["entity_id"] = name
        df["site"] = name

    df["region"] = meta["region"]

    keep_cols = [
        "source_dataset",
        "entity_id",
        "site",
        "region",
        "connect_time_utc",
        "charge_end_time_utc",
        "energy_kwh"
    ]

    df = df[keep_cols]

    out_file = os.path.join(
        OUTPUT_FOLDER,
        file.replace(".csv", "_sessions.csv")
    )

    df.to_csv(out_file, index=False)

    print(df.head())

    print(df.shape)

print("\nDone.")
```

```python
import os
import pandas as pd

from session_to_hourly import (
    sessions_to_hourly,
    validate_split,
    reindex_continuous
)

INPUT_FOLDER = r"harmonized_sessions"
OUTPUT_FOLDER = r"hourly_output"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

for file in os.listdir(INPUT_FOLDER):

    if not file.endswith(".csv"):
        continue

    print(f"\nProcessing {file}")

    sessions = pd.read_csv(
        os.path.join(INPUT_FOLDER, file),
        parse_dates=[
            "connect_time_utc",
            "charge_end_time_utc"
        ]
    )

    hourly = sessions_to_hourly(sessions)

    validate_split(sessions, hourly)

    hourly = reindex_continuous(hourly)

    out = os.path.join(
        OUTPUT_FOLDER,
        file.replace("_sessions.csv", "_hourly.csv")
    )

    hourly.to_csv(out, index=False)

    print(hourly.head())

    print(hourly.shape)

print("\nFinished.")
```

```python
import os
import numpy as np
import pandas as pd
import holidays

LAG_HOURS = [1, 24, 168]
ROLLING_WINDOWS_H = [24, 168]

HOURLY_FOLDER = "hourly_output"
SESSION_FOLDER = "harmonized_sessions"
OUTPUT_FOLDER = "feature_output"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)
```

```python
def add_temporal_features(df):

    df = df.copy()

    df["hour"] = pd.to_datetime(df["hour"], utc=True)

    tz_map = {
        "California": "America/Los_Angeles",
        "Colorado": "America/Denver",
        "France": "Europe/Paris",
        "UK": "Europe/London",
        "Australia": "Australia/Perth",
    }

    holiday_map = {
        "California": holidays.US(),
        "Colorado": holidays.US(),
        "France": holidays.FR(),
        "UK": holidays.UK(),
        "Australia": holidays.AU(),
    }

    local_hour = []
    local_day = []
    local_month = []
    local_date = []

    for _, row in df.iterrows():

        tz = tz_map[row["region"]]

        t = row["hour"].tz_convert(tz)

        local_hour.append(t.hour)
        local_day.append(t.weekday())
        local_month.append(t.month)
        local_date.append(t.date())

    df["hour_of_day"] = local_hour
    df["day_of_week"] = local_day
    df["month"] = local_month

    df["is_weekend"] = (
        df["day_of_week"]
        .isin([5, 6])
        .astype(int)
    )

    df["hour_sin"] = np.sin(
        2 * np.pi * df["hour_of_day"] / 24
    )

    df["hour_cos"] = np.cos(
        2 * np.pi * df["hour_of_day"] / 24
    )

    df["dow_sin"] = np.sin(
        2 * np.pi * df["day_of_week"] / 7
    )

    df["dow_cos"] = np.cos(
        2 * np.pi * df["day_of_week"] / 7
    )

    holidays_list = []

    for region, d in zip(df["region"], local_date):

        holidays_list.append(
            int(d in holiday_map[region])
        )

    df["is_holiday"] = holidays_list

    df["is_paid_period"] = (
        (df["source_dataset"] == "acn")
        &
        (df["site"] == "caltech")
        &
        (
            df["hour"]
            >=
            pd.Timestamp(
                "2018-11-01",
                tz="UTC"
            )
        )
    ).astype(int)

    return df
```

```python
def add_session_count_features(hourly_df, sessions):

    s = sessions.copy()

    s["connect_time_utc"] = pd.to_datetime(
        s["connect_time_utc"],
        utc=True
    )

    s["hour"] = s["connect_time_utc"].dt.floor("h")

    counts = (
        s.groupby(
            ["source_dataset", "entity_id", "hour"]
        )
        .size()
        .reset_index(name="sessions_started")
    )

    df = hourly_df.merge(
        counts,
        on=["source_dataset", "entity_id", "hour"],
        how="left"
    )

    df["sessions_started"] = (
        df["sessions_started"]
        .fillna(0)
        .astype(int)
    )

    shifted = (
        df.groupby(
            ["source_dataset", "entity_id"]
        )["sessions_started"]
        .shift(1)
    )

    df["sessions_started_roll24h"] = (
        shifted.groupby(
            [df["source_dataset"], df["entity_id"]]
        )
        .rolling(
            24,
            min_periods=6
        )
        .mean()
        .reset_index(level=[0, 1], drop=True)
    )

    return df
```

```python
def add_lag_and_rolling_features(df):

    df = df.sort_values(
        ["source_dataset", "entity_id", "hour"]
    ).copy()

    g = df.groupby(
        ["source_dataset", "entity_id"]
    )["kwh_allocated"]

    shifted = g.shift(1)

    # -----------------------
    # Lag Features
    # -----------------------

    for lag in [1, 2, 3, 6, 12, 24, 168]:

        df[f"lag_{lag}h"] = g.shift(lag)

    # -----------------------
    # EMA
    # -----------------------

    df["ema_24h"] = (
        shifted.groupby(
            [df["source_dataset"], df["entity_id"]]
        )
        .transform(
            lambda x: x.ewm(
                span=24,
                min_periods=6
            ).mean()
        )
    )

    # -----------------------
    # Expanding Mean
    # -----------------------

    df["expanding_mean"] = (
        shifted.groupby(
            [df["source_dataset"], df["entity_id"]]
        )
        .transform(
            lambda x: x.expanding(
                min_periods=6
            ).mean()
        )
    )

    # -----------------------
    # Rolling 24h
    # -----------------------

    roll24 = (
        shifted.groupby(
            [df["source_dataset"], df["entity_id"]]
        )
        .rolling(
            24,
            min_periods=6
        )
    )

    df["roll_mean_24h"] = roll24.mean().reset_index(level=[0,1],drop=True)
    df["roll_std_24h"] = roll24.std().reset_index(level=[0,1],drop=True)
    df["roll_median_24h"] = roll24.median().reset_index(level=[0,1],drop=True)
    df["roll_min_24h"] = roll24.min().reset_index(level=[0,1],drop=True)
    df["roll_max_24h"] = roll24.max().reset_index(level=[0,1],drop=True)
    df["roll_q25_24h"] = roll24.quantile(0.25).reset_index(level=[0,1],drop=True)
    df["roll_q75_24h"] = roll24.quantile(0.75).reset_index(level=[0,1],drop=True)

    # -----------------------
    # Rolling 168h
    # -----------------------

    roll168 = (
        shifted.groupby(
            [df["source_dataset"], df["entity_id"]]
        )
        .rolling(
            168,
            min_periods=24
        )
    )

    df["roll_mean_168h"] = roll168.mean().reset_index(level=[0,1],drop=True)
    df["roll_std_168h"] = roll168.std().reset_index(level=[0,1],drop=True)

    # -----------------------
    # Trend
    # -----------------------

    df["trend_1h"] = df["lag_1h"] - df["lag_2h"]

    df["accel_1h"] = (
        df["trend_1h"]
        -
        (df["lag_2h"] - df["lag_3h"])
    )

    df["trend_24h"] = (
        df["roll_mean_24h"]
        -
        df["roll_mean_168h"]
    )

    # -----------------------
    # Previous Peaks
    # -----------------------

    df["prev_day_peak"] = (
        roll24.max()
        .reset_index(level=[0,1],drop=True)
    )

    df["prev_week_peak"] = (
        roll168.max()
        .reset_index(level=[0,1],drop=True)
    )

    # -----------------------
    # Consecutive Zero Hours
    # -----------------------

    zero = shifted.eq(0)

    df["consec_zero_hours"] = (
        zero.groupby(
            [df["source_dataset"], df["entity_id"]]
        )
        .transform(
            lambda x:
            x.groupby(
                (~x).cumsum()
            ).cumsum()
        )
    )

    # -----------------------
    # Consecutive Active Hours
    # -----------------------

    active = shifted.gt(0)

    df["consec_active_hours"] = (
        active.groupby(
            [df["source_dataset"], df["entity_id"]]
        )
        .transform(
            lambda x:
            x.groupby(
                (~x).cumsum()
            ).cumsum()
        )
    )

    # -----------------------
    # Volatility
    # -----------------------

    df["volatility_24h"] = (
        df["roll_std_24h"]
        /
        (df["roll_mean_24h"] + 1e-6)
    )

    return df
```

```python
def build_feature_table(hourly_reindexed, sessions):

    df = add_temporal_features(hourly_reindexed)

    df = add_session_count_features(
        df,
        sessions
    )

    df = add_lag_and_rolling_features(df)

    return df
```

```python
import os
import pandas as pd

HOURLY_FOLDER = "hourly_output"
SESSION_FOLDER = "harmonized_sessions"
OUTPUT_FOLDER = "feature_output"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

hourly_files = sorted(
    f for f in os.listdir(HOURLY_FOLDER)
    if f.endswith("_hourly.csv")
)

for hourly_file in hourly_files:

    dataset = hourly_file.replace("_hourly.csv", "")

    print("=" * 80)
    print("Processing:", dataset)

    hourly = pd.read_csv(
        os.path.join(HOURLY_FOLDER, hourly_file)
    )

    hourly["hour"] = pd.to_datetime(
        hourly["hour"],
        utc=True
    )

    sessions = pd.read_csv(
        os.path.join(
            SESSION_FOLDER,
            dataset + "_sessions.csv"
        )
    )

    sessions["connect_time_utc"] = pd.to_datetime(
        sessions["connect_time_utc"],
        utc=True
    )

    sessions["charge_end_time_utc"] = pd.to_datetime(
        sessions["charge_end_time_utc"],
        utc=True
    )

    features = build_feature_table(
        hourly,
        sessions
    )

    out_file = os.path.join(
        OUTPUT_FOLDER,
        dataset + "_features.csv"
    )

    features.to_csv(
        out_file,
        index=False
    )

    print("Saved:", out_file)
    print("Shape:", features.shape)

print("\nFinished.")
```
